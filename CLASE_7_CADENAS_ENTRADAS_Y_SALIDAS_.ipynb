## El método replace().

```python
comparacion = 'Luis, Hugo, Paco, Luis'.replace("Luis","x").split()
original = 'Luis, Hugo, Paco, Luis'.split()
```

```python
sum([comparacion[x] != original[x] for x in range(0,len(original))])
```
